# Pivot Table Tasks

Applying pivot table operations on User_Data (Sales Dataset) and Titanic Dataset.

```python
import pandas as pd
import numpy as np
```

## Load User Data

```python
df = pd.read_csv('User_Data.csv')
df
```

---
## Task 1
**Write a Pandas program to create a Pivot table and find the region wise, item wise unit sold.**

```python
pivot_task1 = pd.pivot_table(
    df,
    index='Region',
    columns='Item',
    values='Units',
    aggfunc='sum',
    fill_value=0
)
print("Region-wise, Item-wise Units Sold:")
pivot_task1
```

---
## Task 2
**Write a Pandas program to create a Pivot table and find the total sale amount region wise, manager wise, salesman wise where Manager = "Douglas".**

```python
# Filter data for Manager = Douglas
df_douglas = df[df['Manager'] == 'Douglas']

pivot_task2 = pd.pivot_table(
    df_douglas,
    index=['Region', 'Manager', 'SalesMan'],
    values='Sale_amt',
    aggfunc='sum'
)
print("Total Sale Amount - Region wise, Manager wise, Salesman wise (Manager = Douglas):")
pivot_task2
```

---
## Task 3
**Write a Pandas program to create a Pivot table and find manager wise, salesman wise total sale and also display the sum of all sale amount at the bottom.**

```python
pivot_task3 = pd.pivot_table(
    df,
    index=['Manager', 'SalesMan'],
    values='Sale_amt',
    aggfunc='sum',
    margins=True,
    margins_name='Grand Total'
)
print("Manager-wise, Salesman-wise Total Sale Amount (with Grand Total):")
pivot_task3
```

## Pandas Pivot on Titanic Passengers

Loading the Titanic dataset.

```python
titanic = pd.read_csv('titanic.csv')
titanic.head()
```

```python
titanic.info()
print("\nColumn names:", titanic.columns.tolist())
```

---
## Task 4
**Write a Pandas program to create a Pivot table and find survival rate by gender, age of the different categories of various classes.**

```python
# Create age categories
titanic['AgeGroup'] = pd.cut(
    titanic['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child(0-12)', 'Teen(13-18)', 'Young Adult(19-35)', 'Adult(36-60)', 'Senior(60+)']
)

pivot_task4 = pd.pivot_table(
    titanic,
    index=['Sex', 'AgeGroup'],
    columns='Pclass',
    values='Survived',
    aggfunc='mean'
)
pivot_task4.columns = ['Class 1', 'Class 2', 'Class 3']
print("Survival Rate by Gender, Age Group across Classes:")
pivot_task4.round(2)
```

---
## Task 5
**Write a Pandas program to create a Pivot table and find the probability of survival by class, gender, solo boarding and port of embarkation.**

```python
# Create 'SoloBoarding' column: passenger is alone if SibSp + Parch == 0
titanic['SoloBoarding'] = (titanic['SibSp'] + titanic['Parch'] == 0).map({True: 'Solo', False: 'Not Solo'})

pivot_task5 = pd.pivot_table(
    titanic,
    index=['Pclass', 'Sex', 'SoloBoarding'],
    columns='Embarked',
    values='Survived',
    aggfunc='mean'
)
print("Probability of Survival by Class, Gender, Solo Boarding and Port of Embarkation:")
pivot_task5.round(2)
```

---
## Task 6
**Write a Pandas program to create a Pivot table and find number of survivors and average rate grouped by gender and class.**

```python
pivot_task6 = pd.pivot_table(
    titanic,
    index=['Sex', 'Pclass'],
    values='Survived',
    aggfunc=['sum', 'mean']
)
pivot_task6.columns = ['Number of Survivors', 'Survival Rate (Avg)']
pivot_task6['Number of Survivors'] = pivot_task6['Number of Survivors'].astype(int)
print("Number of Survivors and Average Survival Rate by Gender and Class:")
pivot_task6
```

## Task 7
**Write a Pandas program to create a Pivot table and find survival rate by gender, age of the different categories of various classes. Add the fare as a dimension of columns and partition fare column into 2 categories based on the values present in fare columns.**

```python
# Partition Fare into 2 categories based on median
fare_median = titanic['Fare'].median()
print(f"Fare Median: {fare_median:.2f}")

titanic['FareCategory'] = pd.cut(
    titanic['Fare'],
    bins=[titanic['Fare'].min() - 0.01, fare_median, titanic['Fare'].max()],
    labels=['Low Fare', 'High Fare']
)

# Ensure AgeGroup exists (re-create if needed)
titanic['AgeGroup'] = pd.cut(
    titanic['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child(0-12)', 'Teen(13-18)', 'Young Adult(19-35)', 'Adult(36-60)', 'Senior(60+)']
)

pivot_task7 = pd.pivot_table(
    titanic,
    index=['Sex', 'AgeGroup', 'Pclass'],
    columns='FareCategory',
    values='Survived',
    aggfunc='mean'
)
print("\nSurvival Rate by Gender, Age Group, Class with Fare Categories (Low/High):")
pivot_task7.round(2)
```
